```python
import pandas as pd

print("Pandas is working!")
print("Version:", pd.__version__)
```

```python
from pathlib import Path

# Locate the dataset from the project folder
project_folder = Path.cwd()
if project_folder.name == "notebooks":
    project_folder = project_folder.parent

file_path = project_folder / "Data" / "Raw data" / "UNSW_NB15_training-set.csv"

# Load the dataset
df = pd.read_csv(file_path)

# Display dataset dimensions
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python

# Display the first 5 rows of the dataset
df.head()
```

```python

# Inspect dataset structure and data types
df.info()
```

```python
'''
Pseudo code for missing values analysis:
START
    Identify missing values in each column
    Count missing values per column
    Calculate total missing values
    Display results
END
'''
```

```python

# Check missing values per column
missing_values = df.isnull().sum()

print("Missing values per column:")
print(missing_values)

# Calculate total missing values
print("Total missing values:", missing_values.sum())
```

```python
'''
START
    Identify the target variable
    Count records in each class
    Calculate percentages for each class
    Display results
END
'''
```

```python

# Count records per class
class_counts = df["label"].value_counts()

# Calculate percentages
class_percentages = df["label"].value_counts(normalize=True) * 100

print("Class Distribution:")
print(class_counts)

print("\nClass Percentages:")
print(class_percentages.round(2))
```

```python
'''
START
    Locate the testing dataset
    Load the CSV into a new DataFrame
    Check dataset dimensions
    Check missing values
    Check duplicate records
    Check class distribution
END
'''
```

```python

# Locate the testing dataset
test_path = project_folder / "Data" / "Raw data" / "UNSW_NB15_testing-set.csv"

# Load testing dataset
df_test = pd.read_csv(test_path)

# Inspect dataset dimensions
print("Rows:", df_test.shape[0])
print("Columns:", df_test.shape[1])

# Check missing values
print("Missing values:", df_test.isnull().sum().sum())

# Check duplicate records
print("Duplicates:", df_test.drop(columns=["id"]).duplicated().sum())

# Check class distribution
print("\nClass Distribution:")
print(df_test["label"].value_counts())
```

```python
# Final training dataset summary

print("TRAINING DATASET")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicates:", df.drop(columns=["id"]).duplicated().sum())

print("\nClass Distribution:")
print(df["label"].value_counts())
```
